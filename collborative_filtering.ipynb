```python
import os
import numpy
import matplotlib.pyplot as plt
import pandas as pd
plt.rc('font', family='Malgun Gothic')
plt.rcParams['axes.unicode_minus']=False
plt.rcParams['font.family']

ratings = pd.read_csv('../week3/Data_processing/recommend/ratings.csv')
movies = pd.read_csv('../week3/Data_processing/recommend/movies.csv')
print(len(ratings))
print('movie record', ratings['movieId'].nunique())
ratings.head()
```

```python
# user x, movie score array(unseen ;movie score= 0, movie only scores over 50)
popular = ratings.groupby('movieId').size()
popular = popular[popular >= 50].index
r_pop = ratings[ratings['movieId'].isin(popular)]
matrix = (r_pop.merge(movies, on='movieId')
        .pivot_table('rating',index='userId', columns='title')
        .fillna(0))
print('score array:', matrix.shape)
matrix
```

```python
#similarity
from sklearn.metrics.pairwise import cosine_similarity
item_sim = pd.DataFrame(cosine_similarity(matrix.T),
                       index=matrix.columns,
                       columns=matrix.columns)
item_sim
```

```python
def similar_movies(title, n=5):
    return item_sim[title].drop(title).sort_values(ascending=False).head(n)

for t in ['Toy Story (1995)', 'Titanic (1997)']:
    print(f'-------if you like [{t}]------')
    print(similar_movies(t).round(2).to_string())
    print()
```

```python
#user_based
user_sim = pd.DataFrame(cosine_similarity(matrix), 
                        index=matrix.index, columns=matrix.index)
user_sim
```

```python
def recommend_for_user(user_id, n=5, k=10):
    neighbors = (user_sim[user_id].drop(user_id).sort_values(ascending=False).head(k).index) #similar neighbor
    seen = matrix.loc[user_id][matrix.loc[user_id] >0].index #already seen
    scores = matrix.loc[neighbors].mean().drop(seen)
    return scores.sort_values(ascending=False).head(n)
target = matrix.index[0]
print(f'user {target} already seen at high score')
print(matrix.loc[target][matrix.loc[target] >=4].head(4).index.tolist())
print('recommend')
print(recommend_for_user(target).round(2).to_string())
```
